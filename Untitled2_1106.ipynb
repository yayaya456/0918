```python
wall = [45, 61, 27, 38, 50, 74, 63, 19, 56, 62, 82, 36]
answers = []
for i in range(len(wall)):
    for j in range(i + 1, len(wall)):
        if wall[i] + wall[j] == 100:
            answers.append(wall[i])
            answers.append(wall[j])
            break
print("".join(map(str, answers)))
```

```python
n = input("請輸入太空訊號：")
message = []
num_list = n.split()
for num in num_list:
    i = chr(int(num))
    message.append(i)
print(f"解碼結果：{message}")
```

```python
a = {"珍奶": 60, "紅茶": 30, "冬瓜茶": 25}
b = {"雞排": 80, "地瓜球": 40, "烤玉米": 50}
c = []
for a_name,a_price in a.items():
    for b_name,b_price in b.items():
        total_price = a_price + b_price
        if total_price <= 100:
            c.append(f"{a_name}+{b_name}")
print(c)
```
